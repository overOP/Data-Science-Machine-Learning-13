### Math Tutor Using Random, Arithmetic Operators, and OOP

*Scenario:* Create a `MathTutor` class that generates random math questions using `random` and `math operators`.

- Track correct and incorrect answers.
- Provide a score at the end.
- Handle invalid input using exception handling.

> Concepts: Random, Math, Exception Handling, Classes

```python
import random

class MathTutor:
    def __init__(self):
        self.correct = 0
        self.incorrect = 0

    def generate_question(self):
        num1 = random.randint(1, 20)
        num2 = random.randint(1, 20)

        operator = random.choice(["+", "-", "*", "/"])

        if operator == "+":
            answer = num1 + num2
        elif operator == "-":
            answer = num1 - num2
        elif operator == "*":
            answer = num1 * num2
        else:
            num1 = num1 * num2
            answer = num1 // num2

        return num1, num2, operator, answer

    def ask_question(self):
        num1, num2, operator, correct_answer = self.generate_question()

        while True:
            try:
                user_answer = float(
                    input(f"What is {num1} {operator} {num2}? ")
                )

                if user_answer == correct_answer:
                    print("Correct!")
                    self.correct += 1
                else:
                    print(f"Wrong! The answer is {correct_answer}")
                    self.incorrect += 1

                break

            except ValueError:
                print("Invalid input! Please enter a number.")

    def show_score(self):
        total = self.correct + self.incorrect

        if total > 0:
            percentage = (self.correct / total) * 100
        else:
            percentage = 0

        print("\nFinal Score")
        print(f"Correct: {self.correct}")
        print(f"Incorrect: {self.incorrect}")
        print(f"Score: {percentage:.2f}%")

    def start(self):
        print("Welcome to Math Tutor!")
        try:
            questions = int(input("How many questions do you want? "))
            if questions <= 0:
                raise ValueError("Number of questions must be positive.")
            for _ in range(questions):
                self.ask_question()
            self.show_score()
        except ValueError:
            print("Please enter a valid positive number.")

tutor = MathTutor()
tutor.start()
```

### Exception Handling Scenario: Online Age-Restricted Service

**Scenario:** You’re building a sign-up system for an online movie rental platform. Some movies are age-restricted (18+). You need to ensure proper validation and error handling during user registration.

**Task:**

1. Create a custom exception class called `UnderageError` that inherits from `Exception`.
2. Write a function `register_user()` that:
   - Takes a user’s `name` and `age` as input.
   - Raises `UnderageError` if the user is under 18.
   - Otherwise, prints a welcome message.
3. Wrap the function call in a `try` block and handle the exception.
4. Use `else` to confirm successful registration and `finally` to always print **“Thank you for using MovieTime!”** regardless of outcome.

Also try to validate if the age input is numeric. Raise a `ValueError` if not, and handle it separately.

```python
class UnderageError(Exception):
    """Raised when a user is under 18."""
    pass


def register_user(name, age):
    if not age.isdigit():
        raise ValueError("Age must be a numeric value.")

    age = int(age)

    if age < 18:
        raise UnderageError("User must be 18 or older to register.")

    print(f"Welcome, {name}! Your registration was successful.")

name = input("Enter your name: ")
age = input("Enter your age: ")

try:
    register_user(name, age)

except UnderageError as e:
    print(f"Underage Error: {e}")

except ValueError as e:
    print(f"Invalid Input: {e}")

else:
    print("Registration completed successfully!")

finally:
    print("Thank you for using MovieTime!")
```

# Problem Statement

We want to build an online shopping cart system that allows users to add products to their cart, calculate the total cost, apply discounts, and generate an invoice. The system should include the following functionalities:

- Adding products to the cart
- Removing products from the cart
- Calculating the total cost
- Applying discounts based on user type
- Generating an invoice

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} - ${self.price:.2f}"


class User:
    def __init__(self, name, is_admin=False, is_premium=False):
        self.name = name
        self.is_admin = is_admin
        self.is_premium = is_premium
        self.cart = ShoppingCart(self)

    def make_premium(self, user):
        if not self.is_admin:
            print("Only admins can make users premium.")
            return

        user.is_premium = True
        print(f"{user.name} is now a premium member.")

    def create_admin(self, user):
        if not self.is_admin:
            print("Only admins can create other admins.")
            return

        user.is_admin = True
        print(f"{user.name} is now an admin.")

class ShoppingCart:
    def __init__(self, user):
        self.user = user
        self.products = []

    def add_product(self, product):
        self.products.append(product)
        print(f"{product.name} added to {self.user.name}'s cart.")

    def remove_product(self, product_name):
        for product in self.products:
            if product.name.lower() == product_name.lower():
                self.products.remove(product)
                print(f"{product.name} removed from cart.")
                return

        print("Product not found in cart.")

    def calculate_total_cost(self):
        total = sum(product.price for product in self.products)

        if self.user.is_premium:
            discount = total * 0.10
            total -= discount

        return total

    def generate_invoice(self):
        print("INVOICE")

        print(f"Customer: {self.user.name}")
        print(f"Premium: {'Yes' if self.user.is_premium else 'No'}")

        if not self.products:
            print("Cart is empty.")
            return

        subtotal = sum(product.price for product in self.products)

        for product in self.products:
            print(f"{product.name:<25} ${product.price:.2f}")

        print(f"Subtotal:${subtotal:.2f}")

        if self.user.is_premium:
            discount = subtotal * 0.10
            print(f"Premium Discount (10%): -${discount:.2f}")
        else:
            discount = 0

        total = subtotal - discount

        print(f"Total: ${total:.2f}")

laptop = Product("Laptop", 1000)
phone = Product("Smartphone", 700)
headphones = Product("Headphones", 100)

admin = User("Admin", is_admin=True)
john = User("John")
alice = User("Alice")

admin.make_premium(john)
admin.create_admin(alice)
john.make_premium(alice)
john.cart.add_product(laptop)
john.cart.add_product(headphones)

print(f"\nJohn's total: ${john.cart.calculate_total_cost():.2f}")

john.cart.remove_product("Headphones")

print(f"John's total after removal: "
      f"${john.cart.calculate_total_cost():.2f}")

john.cart.generate_invoice()
alice.cart.add_product(phone)

print(f"\nAlice's total: ${alice.cart.calculate_total_cost():.2f}")

alice.cart.generate_invoice()
admin.cart.add_product(headphones)

print(f"\nAdmin's total: ${admin.cart.calculate_total_cost():.2f}")

admin.cart.generate_invoice()
```
