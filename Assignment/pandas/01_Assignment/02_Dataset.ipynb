### **Problem 3: Handling Missing Data and Imputation**  
1. Identify columns with **missing values**.
2. Impute the missing values:
   - For **numerical columns**, replace missing values with the **median**.
   - For **categorical columns**, replace missing values with the **most frequent** value.
3. Justify your choice of imputation methods and check if any missing data remains.

```python
#1. Identify columns with missing values
missing_values = df.isna().sum()
missing_values[missing_values > 0]
```

```python
#2. Impute missing values
numerical_columns = df.select_dtypes(include="number").columns
categorical_columns = df.select_dtypes(include="object").columns

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
#3. Check if missing data remains
missing_after = df.isnull().sum()
missing_after[missing_after > 0]
```

### **Problem 4: Numpy Array Operations**  
1. Convert the numerical columns (such as `LotArea`, `YearBuilt`, `GrLivArea`, `SalePrice`, etc.) into a **NumPy array**.
2. Perform the following NumPy operations on the array:
   - Compute the **mean** and **standard deviation** of each column.
   - Normalize the data using **min-max scaling** and **z-score normalization**.
   - Compute the **dot product** between two numerical columns of your choice.

```python
# 1. Convert numerical columns to NumPy array
numerical_columns = df.select_dtypes(include="number").columns
num_array = df[numerical_columns].to_numpy()

print("Array shape:", num_array.shape)
```

```python
# 2. Mean and standard deviation
mean = np.mean(num_array, axis=0)
std = np.std(num_array, axis=0)

for column, m, s in zip(numerical_columns, mean, std):
    print(f"{column}: Mean = {m:.2f}, Std = {s:.2f}")
```

```python
# 3. Min-Max Scaling
min_values = np.min(num_array, axis=0)
max_values = np.max(num_array, axis=0)

min_max_scaled = (
    (num_array - min_values)
    / (max_values - min_values)
)

print("\nMin-Max Scaled Data:")
print(min_max_scaled)
```

```python
# 4. Z-score Normalization
z_score = (
    (num_array - mean)
    / std
)

print("\nZ-Score Normalized Data:")
print(z_score)
```

```python
# 5. Dot Product
gr_liv_area = df["GrLivArea"].to_numpy()
sale_price = df["SalePrice"].to_numpy()

dot_product = np.dot(gr_liv_area, sale_price)

print("\nDot Product:", dot_product)
```

### **Problem 6: Aggregation and Transformation**  
1. Perform a **groupby** operation on `Neighborhood` and calculate the **mean** and **median** for `SalePrice`.
2. Use the `apply()` function to create a custom transformation for `SalePrice` (e.g., scaling or transformation).
3. Aggregate multiple statistical values (e.g., **mean**, **median**, and **count**) for `SalePrice` grouped by `OverallQual`.

```python
# 1. Group by Neighborhood
neighborhood_prices = df.groupby("Neighborhood")["SalePrice"].agg(
    ["mean", "median"]
)

neighborhood_prices.columns = [
    "Mean_SalePrice",
    "Median_SalePrice"
]

print("Sale Price by Neighborhood:")
print(neighborhood_prices)
```

```python
# 2. Custom transformation using apply()
min_price = df["SalePrice"].min()
max_price = df["SalePrice"].max()

df["SalePrice_Scaled"] = df["SalePrice"].apply(
    lambda price: (price - min_price) / (max_price - min_price)
)

df[["SalePrice", "SalePrice_Scaled"]].head()
```

```python
# 3. Aggregate SalePrice by OverallQual
quality_summary = df.groupby("OverallQual")["SalePrice"].agg(
    ["mean", "median", "count"]
)

quality_summary.columns = [
    "Mean_SalePrice",
    "Median_SalePrice",
    "House_Count"
]

print("\nSale Price by Overall Quality:")
print(quality_summary)
```
