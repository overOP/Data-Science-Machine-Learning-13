```python
import pickle
import numpy as np

from tensorflow.keras.models import load_model
```

```python
model = load_model("../data/iris_classifier.keras")
```

```python
model
```

```python
with open("../data/iris_classifier.pkl", "rb") as file:
    objects = pickle.load(file)
```

```python
scaler = objects['scaler']
encoder = objects['encoder']
pkl_model = objects["model"]
```

```python
# User Input
my_data = np.array([
    [12.6, 2.6, 8.5, 0.9],
    [6.1, 3.0, 4.9, 1.5]
])
```

```python
classes = encoder.categories_[0]

# Pre-Processing
scaled_data = scaler.transform(my_data)

# Predict
predict = np.argmax(
    model.predict(scaled_data),
    axis= 1
)

[classes[res] for res in predict]
```
