# Markov Decision Processes

In the last chapter, we learned about Markov Processes and Markov Reward Processes (MRPs). With MRPs in particular, we saw that we could evaluate the value of states given that we know the reward function and transition matrix for the process. Something crucial is missing so far though: we have no way to act in environments. Evaluating an environment is useful, but we can only go so far if we cannot take actions to influence the states that we visit, or the rewards that we get. 

In this chapter, we will extend MRPs to include actions. This will give rise to a process called a Markov Decision Process. 


Markov Decision Process is a Markov Reward Process with action added to it. A Markov Decision Process is defined by the tuple, $(S, A, P, R, \gamma)$
>* $S$ is a finite set of states
>* $A$ is a finite set of actions
>* $P$ is a state transition probability matrix,
$$p^a _{ss'} = P[S_{t+1} = s' | S_t = s, A_t =a]$$
>* $R$ is a reward function, 
$$R_s ^a = E[R_{t+1}| S_t = s, A_t =a]$$
>* $\gamma$ is a discount factor, $\gamma \in [0, 1]$



In an MDP, the value of a state does not just depend on what the future states are like, it also depends on the actions we take in each state. That is, we can influence where we end up, and the rewards we get by acting accordingly in the environment. 


The way we act in an environment is governed by a function called our policy. It is a function that takes in the current state we are in, and gives out the action we might take. The output from a policy can be the action to take, or a probability distribution over all the possible actions. Generally speaking, we can define the policy as a probability distribution from states to actions. A policy is a simple function, that defines a probability distribution over action $(a \in A)$ for each state $(s \in S)$. If an agent follows a policy $'\pi'$ then $\pi (a | s)$ is the probability that the agent takes the action $a$ at the state $s$.


$$\pi (a | s) = P[A_t = a | S_t = s]$$


### Value functions for an MDP

In an MDP, we are interested in two kinds of value functions. The first one is the state value function, which we have looked at in the previous chapter. The second one, is the action value function. 

1. State value function $V_\pi(s)$

    The value function of a state $s$ under a policy $\pi$ , denoted by $v_\pi(s)$, is the expected return when starting in $s$ and following $\pi$ thereafter. So, it is the average return you get when you start in the state $s$, and then follow the policy $\pi$.

    The state value function $v_\pi (s)$ of an MDP is the expected return starting from state $s$ and then following policy $\pi$. It calculates how much reward an agent gets by following the policy. Simply, it gives how good it is to be in the state $s$ by following the policy $\pi$.

    Mathematically:
$$v_\pi(s) = \mathbb{E}_{\pi}[\sum_{k=0}^{\infty}\gamma^kR_{t+k+1}|S_t=s], for\ all\ s\ \epsilon\ \mathbb{S}$$

$$v_\pi(s) = \mathbb{E}_\pi [G_t | S_t = s]$$


2. Action value function $q_\pi(s, a)$

    Similarly, we define the value of taking action $a$ in state $s$ under a policy $\pi$, denoted as $q_{\pi}(s,a)$, as the expected return starting from $s$, taking the action $a$, and thereafter following policy $\pi$:

    $$q_\pi (s, a) = \mathbb E_\pi (G_t | S_t = s, A_t = a)$$

    Basically, it tells us the value of performing a certain action($a$) in a state($s$) with a policy $\pi$, i.e., how good is it to take the particular action. 


## Bellman Equations for MDPs

Like the Bellman equations for MRPs we saw earlier, we can formulate a recursive relation of the state value functions and the action value functions for MDPs. 


The state value function can be decomposed into immediate reward plus discounted value of successor state.
$$v_\pi(s) = \mathbb E _\pi[R_{t+1} + \gamma V_\pi (S_{t+1}) | S_t = s]$$

The action value can similarly be decomposed as:
$$q_\pi (s, a) = \mathbb E_\pi [R_{t+1} + \gamma \space q_\pi(S_{t+1}, A_{t+1}) | S_t = s, A_t = a ]$$

The white circle is a general state, and the black circles represents actions you can take from the state. The value function of a state $s$ can be written as:

$$v_\pi(s) = \sum _{a \in A} \pi(a|s) q_\pi (s, a)\quad ... \quad (1)$$

Each action has an action value associated with it, and the value of a state is the average return you get from these actions. Since we do not take all actions uniformly, and are sampling actions from the policy, we average the values using the policy $\pi$.


Also, the action values can be written in terms of the state values. 


The value of an action $a$ in a state $s$ is the reward $r$ you get when you take that action, and the average value of the states you end up at in the next time step. 

$$q_\pi (s, a) = R_s ^a + \gamma \sum _{s' \in S} p^a _{ss'} V_\pi (s')$$


If we combine the two equations, we get the following:

$$v_\pi (s) = \sum _{a \in A} \pi(a|s) (R^a _s + \gamma \sum _{s' \in S}p^a _{ss'} v_\pi (s'))$$


To understand this, lets look at the backup diagram for this equation.


We have written the value function of a state $s$ as a linear combination of the value of all the states in the MDP. The value of a state is given by the sum of the average immediate reward we get from that state ($\sum_a \pi(a|s) * R^a_s$) and the average value of the states that we can end up at ($\pi(a|s) * \gamma \sum _{s'}p^a _{ss'} v_\pi (s')$). 


We can do the same for the action value function as well, which gives us the following equation.

$$q_\pi (s, a) = R_s ^a + \gamma \sum_{s' \in S} p^a _{ss'} \sum _{a' \in A} \pi(a'|s')q_\pi(s', a')$$


These equations are also called the Bellman expectation equations. Again, these equations can be written using vectors and matrices. The notation has been simplified for clarity. 
$$
v_\pi=R_\pi + \gamma P_\pi v_\pi
$$

And like we did before, we can solve these equations by isolating the $v_\pi$ vector.


$$
v_\pi = (I−γP_\pi)^{−1}R_\pi
$$


## Key Takeaways

1. Markov Decision Processes are Markov Processes in which you can take actions, and these actions result in some rewards. 

2. Both states and actions have values associated with them. The value of a state gives how good that state is under a policy. The value of an action tells us how good it is to take that action, and then follow our policy from then on.

3. Both state and action value functions give rise to their own form of Bellman equations, which we can solve to determine the value functions.
