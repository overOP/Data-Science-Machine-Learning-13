# Pre-Processing

```python
X = data.drop('Outcome', axis= 1).to_numpy()
y = data['Outcome'].to_numpy()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size= 0.2,
    random_state= 909
)
```

```python
log_reg = LogisticRegression(max_iter = 500)
```

```python
log_reg.fit(
    X_train,
    y_train
)
```

```python
y_pred = log_reg.predict(X_test)
```

```python
log_reg.score(
    X_test,
    y_test
)
```
