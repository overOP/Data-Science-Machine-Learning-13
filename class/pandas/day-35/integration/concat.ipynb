```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'name': ['Ram', 'Shyam', 'Hari', 'Krishna'],
    'age': [21, 65, 34, 88]
})

df2 = pd.DataFrame({
    "id" : [1, 2, 3, 5], 
    "principle": [500, 658, 4032, 500], 
    "interest" : [6.3, 4.25, 9.8, 5.6]
})
```

```python
print(df1, df2)
```

## Concatenation

Concatenation is the process of combining two or more dataframes by appending them along a particular axis. To concatenate two or more dataframes along rows or columns, you can use the `pd.concat()` function from the `pandas` library.

```python
# Basic
pd.concat(
    [df1,df2]
)
```

```python
# ignore_index = True
pd.concat(
    [df1, df2],
    ignore_index= True
)
```

```python
# Horizontal Concatenation
pd.concat(
    [df1, df2],
    axis= 1
)
```

```python
# Different Number of Rows
data1 = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Marks": [85, 62, 91]
})

data2 = pd.DataFrame({
    "City": ["Kathmandu", "Pokhara"]
})

pd.concat(
    [data1, data2],
    axis= 1
)
```

```python
# Different Columns
data1 = pd.DataFrame({
    "Name": ["Alice", "Bob"],
    "Marks": [85, 62]
})

data2 = pd.DataFrame({
    "Name": ["Charlie", "David"],
    "Age": [19, 22]
})

pd.concat(
    [data1, data2],
    axis= 0,
    ignore_index= True
)
```

```python
# join

data1 = pd.DataFrame({
    "Name": ["Alice", "Bob"],
    "Marks": [85, 62]
})

data2 = pd.DataFrame({
    "Name": ["Charlie", "David"],
    "Age": [19, 22]
})

pd.concat(
    [data1, data2],
    axis= 0,
    join= "inner",
    ignore_index= True
)
```

```python
# join="outer" vs join="inner"

data1 = pd.DataFrame({
    "Name": ["Alice", "Bob"],
    "Marks": [85, 62],
    "City": ["Kathmandu", "Pokhara"]
})

data2 = pd.DataFrame({
    "Name": ["Charlie", "David"],
    "Age": [19, 22],
    "City": ["Lalitpur", "Bhaktapur"]
})

print(pd.concat(
    [data1, data2],
    join= "outer",
    ignore_index= True
))

print("-" * 50)

print(pd.concat(
    [data1, data2],
    join= "inner",
    ignore_index= True
))
```
