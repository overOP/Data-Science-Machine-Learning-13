## Handling Null or NaN or NA values

```python
df.isna().sum()
```

1. Removing Null Values 

2. Fill something (Imputation)

3. Null as new feature (Feature Engineering)

## 0 -> column wise -> vertical

## 1 -> row wise -> horizontal

```python
# Removing Null Values
df.dropna(subset= "Embarked", axis= 0, ignore_index= True, inplace= True)
```

```python
df.shape
```

```python
# Null value as feature
df.Cabin = df.Cabin.notna().astype(int)
```

```python
df.isna().sum()
```

```python
df.head()
```

```python
# Fill Null Values

age_median = df.Age.median()
age_median
```

```python
df.Age.plot(
    kind= "hist",
    bins= 20,
    edgecolor= "black"
)
```

```python
df_fill_Age = df.copy()
df_fill_Age.loc[:, "Age"] = df_fill_Age.Age.fillna(
    value= age_median,
    inplace= False
)
```

```python
df_fill_Age.isna().sum(axis= 0)
```

```python
df_fill_Age.Age.describe()
```

```python
df_fill_Age.Age.plot(
    kind= "hist",
    bins= 20,
    edgecolor= "black"
)
```

```python
group_by_sex = df.groupby(by= "Sex")
```

```python
median_ages = group_by_sex["Age"].transform("median")
```

```python
df_fill_Age_with_sex = df.copy()

df_fill_Age_with_sex.loc[:, "Age"] = df_fill_Age_with_sex.Age.fillna(
    value= median_ages
)
```

```python
df_fill_Age_with_sex.isna().sum()
```

```python
df_fill_Age_with_sex.Age.plot(
    kind= "hist",
    bins= 20,
    edgecolor= "black"
)
```

## Handling Null value for Age, using initials groups.

```python
group_by_initials = df.groupby(
    by= "initials"
)
```

```python
median_ages = group_by_initials["Age"].transform("median")
```

```python
df_fillAge_with_initial = df.copy()

df_fillAge_with_initial.loc[:, "Age"] = df_fillAge_with_initial.Age.fillna(
    value= median_ages
)
```

```python
df_fillAge_with_initial.Age.plot(
    kind= "hist",
    bins= 20,
    edgecolor= "black"
)
```
