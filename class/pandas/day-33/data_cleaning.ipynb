# Handling Incorrect Data

```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame({
    'id': [1, 3, 5, 7, 9],
    'name': ['John', 'Jane', 'Jennifer', 'Johan', 'Jeff'],
    'age': [25, -32, 19, 42, '55'],
    'gender': ["M", "M", "Z", "F", "F"],
    'income': [400, 700, 300, np.nan, 600]
})

data
```

```python
data.info()
data.describe(include='all')
```

```python
data.isna().sum()
```

```python
data.dtypes
```

```python
# Negative Age
data = data.astype({
    'age': 'int32'
})

data.age = abs(data.age)
data
```

```python
# Incorrect Gender
data = data[data.gender.isin(["M", "F"])]
```

```python
# Fill na values
mean = round(data.income.mean(), 1)
data.income = data.income.fillna(mean)
```
