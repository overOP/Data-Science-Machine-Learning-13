```python
import pandas as pd
import numpy as np
import seaborn as sns
df = pd.read_csv("../data/titanic_data.csv")
```

```python
df.head()
```

### 7.Box Plot

```python
df[["Age", "Fare"]].describe()
```

```python
ax = df.plot(
    kind= "box",
    column= "Age",
    vert=False,
    figsize= (10, 5),
    xticks= range(0, 85, 3),
    grid= True,
)
ax.set_title("Boxplot of Age")
ax.set_xlabel("Age")
ax.set_ylabel("Distribution")
ax.set_yticklabels(["Age"])
```

```python
ax = df.plot(
    kind= "box",
    column= "Fare",
    vert= False,
    figsize= (10, 5),
    grid= True
)
```

```python
ax = df.plot(
    kind= "box",
    column= "Fare",
    by= "Sex",
    vert= False,
    figsize= (12, 3),
    xticks= range(0, 600, 50),
    grid= True
)
```

### 7.Heat Map

```python
correlation = df.corr(numeric_only=True)
correlation
```

```python
sns.heatmap(correlation, annot=True, cmap="coolwarm")
```

```python
mask = np.triu(np.ones_like(correlation, dtype= bool))
```

```python
ax = sns.heatmap(correlation, cmap="coolwarm", fmt=".2f", annot= True, mask= mask)
```

```python
tabel = pd.crosstab(
    df.Survived, df.Pclass, margins= True
)
ax = sns.heatmap(
    data= tabel,
    cmap= "coolwarm",
    fmt= ".1f",
    annot= True,
    linewidths= 0.5,
)
```

```python
# Seaborn
# bubble chart

ax = sns.relplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Sex",
    size="SibSp",
    alpha=0.5,
    height=5,
)
```

```python
# violin plot
sns.violinplot(
    data= df, 
    y = "Age",
    x = "Sex",
    hue = "Survived",
    inner = "quart",
    orient = "v",
    split = True
)
```

```python
# violin plot

ax = sns.violinplot(
    data=df,
    x="Pclass",
    y="Fare",
    hue="Sex",
    split=True,
    inner="quartile",
    palette="Set2",
)
```
