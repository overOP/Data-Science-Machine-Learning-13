## Conditional Based Filtering

```python
mask = df.Fare > 100
fare_100 = df[mask]
fare_100.shape
```

Out of 891 passangers , only `53` passangers paid more than `100$` in fare.

```python
female_fare_100 = fare_100[fare_100.Sex == "female"]
female_fare_100.shape
```

```python
print(f"Out of 53, {round((34/53) * 100, 2)} % are female passengers")
```

```python
male_fare_100 = df[
    (df.Fare > 100) & (df.Sex == "male")
]
male_fare_100.shape
```
