## Reading Data From Excel Files

- `pd.read_excel(file_path, sheet_name=0)`: Reads an Excel file and returns a Pandas DataFrame
  - *file_path*: Path to the Excel file
  - *sheet_name*: Name or index of the Excel sheet to read (default is `0`, meaning the first sheet)

```python
import pandas as pd
```

```python
sc_df = pd.read_excel("../data/Supply_Chain.xlsx", sheet_name=0)
sc_df.shape
```
