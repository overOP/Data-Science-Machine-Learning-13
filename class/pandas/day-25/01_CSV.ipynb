## Datatype of Pandas
1. DataFrame: 2D Table with rows and columns
2. Series: 1D array-like , can hold any data type such as integers, strings, floating-point numbers, and Python objescts

## Creating DataFrame

```python
data = {
    "name": ["Bob", "Aman", "Charls", "Jhon"],
    "age": [25, 56, 15, 78]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
my_data = {
    "id": 1,
    "name": "Pradeep",
    "age": 25
}

res = pd.DataFrame(data=my_data, index=["a", "b", "c"], columns=["id", "name", "age"])
res
```

## Creating Series

```python
a = ["g", "h", "i", "j"]
res = pd.Series(a)
res
```

```python
a = ["1", "2", "3", "4"]
res = pd.Series(data= a, index= ["a", "b", "c", "d"], dtype= "int8", name= "list of number", copy= False)
res
```

```python

my_data = {
    "id": 1,
    "name": "pradeep",
    "age": 25,
}
res_dict = pd.Series(data=my_data)
res_dict
     
```

## Reading Data From Files

* `pd.read_csv(file_path, delimiter=",")`: Reads CSV (Comma Separated Value) file and returns a Pandas DataFrame 
* file_path: Path to csv file
* delimiter: Symbol separating columns (default= ",")

```python
titanic_df = pd.read_csv("../data/titanic_data.csv", delimiter=",", header=0)
titanic_df.shape
```
