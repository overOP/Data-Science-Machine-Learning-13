List Methods

```python
fruits = ["apple", "banana"]
fruits.append("mango")
print(fruits)
```

```python
fruits = ["apple", "banana"]
fruits.extend(["mango", "orange"])
print(fruits)
```

```python
fruits = ["apple", "mango"]
fruits.insert(1, "banana")
print(fruits)
```

```python
numbers = [10, 20, 30, 20]
numbers.remove(20)
print(numbers)
```

```python
numbers = [10, 20, 30]
removed = numbers.pop()
print(removed)
print(numbers)
```

```python
numbers = [1, 2, 3]
numbers.clear()
print(numbers)
```

```python
fruits = ["apple", "banana", "mango"]
print(fruits.index("banana"))
```

```python
numbers = [1, 2, 2, 3, 2]
print(numbers.count(2))
```

```python
numbers = [5, 1, 4, 2]
numbers.sort()
print(numbers)
```

```python
numbers = [1, 2, 3]
numbers.reverse()
print(numbers)
```

```python
numbers = [1, 2, 3]
new_list = numbers.copy()

print(new_list)
```
