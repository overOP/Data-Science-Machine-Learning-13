```python
numbers = {1, 2, 3}

numbers.add(4)

print(numbers)
```

```python
numbers = {1, 2, 3}

numbers.update([4, 5, 6])

print(numbers)
```

```python
numbers = {1, 2, 3}

numbers.remove(2) # if not present, it will throw an error

print(numbers)
```

```python
numbers = {1, 2, 3}

numbers.discard(5) # if not present, it will not throw an error

print(numbers)
```
