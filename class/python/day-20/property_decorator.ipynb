# Python Property Decorator

In Python, the `@property` decorator is used to define **getter, setter, and deleter methods** for a class attribute. These methods allow you to access and modify attributes in a controlled manner.

## Getter

A **getter method** is used to get the value of an attribute. It is called automatically when the attribute is accessed.

In Python, we can define a getter method using the `@property` decorator.

### Example

```python
class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name


person = Person("Pradeep")

print(person.name)

````python
class Person:
    def __init__(self, name):
        self.__full_name = name

    @property
    def name(self): #<- Getter Method
        print("Getting Name")
        return self.__full_name
````

````python
obj = Person("Ram Pd. Luitel")
````

````python
obj.name
````

# Python Property Decorator

In Python, the `@property` decorator is used to define **getter, setter, and deleter methods** for a class attribute. It allows attributes to be accessed and modified in a controlled manner.

## Getter

A **getter method** is used to get the value of an attribute. It is called automatically when the attribute is accessed.

In Python, we can define a getter method using the `@property` decorator.

### Example

```python
class Person:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name


person = Person("Pradeep")

print(person.name)

````python
import uuid

class Person:
    def __init__(self, name):
        self.__full_name = name

    @property
    def user_id(self): #<- Getter Method
        print("Getting ID")
        return f"{self.__full_name[-5:]}_{str(uuid.uuid1())[:8]}"
````

## Setter

A **setter method** is used to set or modify the value of an attribute. It is called automatically when a new value is assigned to the attribute.

In Python, we can define a setter method using the `@property` decorator together with the `@<property_name>.setter` decorator.

### Example

```python
class Person:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        self._name = new_name


person = Person("Pradeep")

print(person.name)

person.name = "John"

print(person.name)

````python
class Person:
    def __init__(self, name):
        self.__full_name = name

    @property
    def name(self): #<- Getter Method
        print("Getting Name")
        return self.__full_name

    @name.setter
    def name(self, value): #<- Setter Method
        print("setting Name")
        self.__full_name = value
````

````python

person = Person("Jack Sparrow")
````

````python
person.name
````

````python
person.name = "Rajesh Hamal"
````

````python
person.name
````

## Deleter

A **deleter method** is used to delete an attribute. It is called automatically when the `del` statement is used to delete the attribute.

In Python, we can define a deleter method using the `@property` decorator together with the `@<property_name>.deleter` decorator.

### Example

```python
class Person:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.deleter
    def name(self):
        del self._name


person = Person("Pradeep")

print(person.name)

del person.name

````python
class Person:
    def __init__(self, name):
        self.__full_name = name

    @property
    def name(self): #<- Getter Method
        print("Getting Name")
        return self.__full_name

    @name.setter
    def name(self, value): #<- Setter Method
        print("setting Name")
        self.__full_name = value

    @name.deleter
    def name(self):
        print("Deleting Name")
        self.__full_name = ""
        # del self.__full_name
````

````python
person = Person("Anmol KC")
````

````python
person.name
````

````python
person.name = "The Rock"
````

````python
person.name
````

````python
del person.name
````

````python
person.name
````
