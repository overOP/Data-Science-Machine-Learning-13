Dictionary Methods

```python
student = {
    "name": "Pradeep",
    "age": 22,
    "course": "Python"
}

print(student.keys())
```

```python
student = {
    "name": "Pradeep",
    "age": 22,
    "course": "Python"
}

print(student.values())
```

```python
student = {
    "name": "Pradeep",
    "age": 22
}

print(student.items())
```

```python
student = {
    "name": "Pradeep",
    "age": 22
}

student.update({"age": 23, "city": "Kathmandu"})

print(student)
```

```python
student = {
    "name": "Pradeep",
    "age": 22
}

age = student.pop("age")

print(age)
print(student)
```

```python
student = {
    "name": "Pradeep",
    "age": 22
}

print(student.popitem())
print(student)
```

```python
student = {
    "name": "Pradeep",
    "age": 22
}

student.clear()

print(student)
```

```python
student = {
    "name": "Pradeep",
    "age": 22
}

new_student = student.copy()

print(new_student)
```

```python
student = {
    "name": "Pradeep"
}

student.setdefault("age", 22)

print(student)
```

```python
keys = ("name", "age", "city")

student = dict.fromkeys(keys, "Unknown")

print(student)
```

Dict sort

```python
student = {
    "c": 3,
    "a": 1,
    "b": 2
}

sorted_dict = dict(sorted(student.items()))

print(sorted_dict)
```

```python
student = {
    "c": 3,
    "a": 1,
    "b": 2
}

sorted_dict = dict(sorted(student.items(), reverse=True))

print(sorted_dict)
```

```python
student = {
    "c": 3,
    "a": 1,
    "b": 2
}

print(sorted(student.keys()))
```

```python
student = {
    "Ram": 90,
    "Shyam": 80,
    "Hari": 95
}

print(sorted(student.values()))
```
