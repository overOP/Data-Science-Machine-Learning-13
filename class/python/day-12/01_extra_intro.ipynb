## Types of Attributes

#### 1. Instance Attributes

These are the attributes that belong to instances of a class.

They are defined within the constructor method `__init__` and can be accessed using the `self` keyword, like `self.name` and `self.age` in the `Dog` class.

They are initialized when a new instance of the class is created.

#### 2. Class Attributes

Class attributes are attributes that belong to the class itself.

They are defined outside the constructor method `__init__` and can be accessed using the class name.

Class attributes are shared by all instances of the class, like the `count` attribute in the `Dog` class.

```python
class Dog:
    count = 0 #<- Class Attribute

    def __init__(self, name: str, age: int = 1):
        # Initializer or Constructor
        self.name = name #<- instance attribute
        self.age = age #<- instance attribute

        Dog.count += 1

dog1 = Dog("Luffy", 3)
print(dog1.name, dog1.age)
dog1.count
```

```python
dog2 = Dog("Simba", 4)
print(dog2.name, dog2.age)
print(dog2.count)
```

## Types of Attributes

#### 1. Instance Attributes

These are the attributes that belong to instances of a class.

They are defined within the constructor method `__init__` and can be accessed using the `self` keyword, like `self.name` and `self.age` in the above `Dog` class.

They are initialized when a new instance of the class is created.

#### 2. Class Attributes

Class attributes are attributes that belong to the class itself.

They are defined outside the constructor method `__init__` and can be accessed using the class name.

Class attributes are shared by all instances of the class, like the `count` attribute in the `Dog` class below.

```python
class Person:
    count = 0 # <- Class attribute

    def __init__(self, name: str, address: str = "kathmandu"):
        # Initializer or Constructor
        self.name = name # <- Instance attribute
        self.address = address # <- Instance attribute

        Person.count += 1

    def get_details(self): # <- Instance method
        return f"I am {self.name} from {self.address}"

    @classmethod
    def get_count(cls): # <- Class Meethod
        return cls.count

    @staticmethod
    def get_full_name(first, last): # <- Static Method
        return f"{last, {first}}"

student1 = Person("Breandon", "Norway")
print(student1.get_details())
print(student1.get_count())
print(student1.get_full_name("Brandon", "Heffies"))
```

### Calculating Student Results

Develop a class to accept a student's name and marks in three subjects, then calculate and display the total and average marks.

### Methods

- `total()` → Calculate the total marks.
- `average()` → Calculate the average marks.
- `display()` → Display the student's name, total marks, and average marks.

```python
class StudentResult:

    def __init__(self, name: str, *marks: int):
        self.name = name
        self.marks = marks

    def total(self):
        return sum(self.marks)

    def average(self):
        all_total = self.total()
        return all_total / len(self.marks)

    def display(self):
        print(f"Name: {self.name}")
        print("-" * 50)
        print(f"Total Marks: {self.total()}")
        print(f"Average Marks: {self.average()}")

bigyan = StudentResult("Pradeep Chaudhary", 90, 34, 56, 56, 89)
print(bigyan.display())
```
