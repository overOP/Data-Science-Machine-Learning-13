# Python Decorators

In Python, a **decorator** is a special type of function that is used to **modify or extend the behavior of another function** without changing its original code.

Decorators are represented by the `@` symbol followed by the name of the decorator function. They are placed immediately before the function that they modify.

### Basic Syntax

```python
def decorator_function(func):
    def wrapper():
        # additional behavior
        func()
    return wrapper


@decorator_function
def my_function():
    print("Hello, Python!")
```

Here, `@decorator_function` applies the decorator to `my_function`.

The following:

```python
@decorator_function
def my_function():
    print("Hello, Python!")
```

is equivalent to:

```python
def my_function():
    print("Hello, Python!")

my_function = decorator_function(my_function)
```

### Why Use Decorators?

Decorators are useful when you want to add common functionality to multiple functions, such as:

* Logging
* Measuring execution time
* Authentication and authorization
* Input validation
* Caching
* Access control

### Implementation

````python
def decorator(func): #<- Function as argument/paremater
    
    # Inner Function
    def wrapper():
        
        print("Before Function Call")
        print(func().upper()) #<- actual function call
        print("After Function Call")
        
    return wrapper #<- return function as value
````

````python
@decorator
def greet():
    return "Hello, World"
````

````python
def validate_email(func):
    """
    Validate email before function run
    """
    def inner(user_email:str):
        if user_email.endswith("@skill.com"):
            return func(user_email)
            
        return "Only support email from SkillShikshya"

    return inner
````

````python
import time

def timeit(func):
    """
    total time taken by function run
    """
    def inner(*args, **kwargs):
        start = time.time()
        res = func(*args, **kwargs)
        
        print("Total time taken", int(start - time.time()) , "seconds")
        return res

    return inner
````
