## Format Code List

The table below shows the commonly used format codes that you can pass to the `strftime()` method.

> **Note:** The `%-` directives (such as `%-d`, `%-m`, `%-Y`) are supported on Unix/Linux/macOS systems but are not available on Windows. For portable Python code, avoid relying on them.

| **Directive** | **Meaning** | **Example** |
|---|---|---|
| `%a` | Abbreviated weekday name. | Sun, Mon, ... |
| `%A` | Full weekday name. | Sunday, Monday, ... |
| `%w` | Weekday as a decimal number. Sunday is `0` and Saturday is `6`. | 0, 1, ..., 6 |
| `%d` | Day of the month as a zero-padded decimal number. | 01, 02, ..., 31 |
| `%-d` | Day of the month as a decimal number without zero-padding. | 1, 2, ..., 31 |
| `%b` | Abbreviated month name. | Jan, Feb, ..., Dec |
| `%B` | Full month name. | January, February, ... |
| `%m` | Month as a zero-padded decimal number. | 01, 02, ..., 12 |
| `%-m` | Month as a decimal number without zero-padding. | 1, 2, ..., 12 |
| `%y` | Year without century as a zero-padded decimal number. | 00, 01, ..., 99 |
| `%-y` | Year without century as a decimal number without zero-padding. | 0, 1, ..., 99 |
| `%Y` | Year with century as a decimal number. | 2013, 2019, ... |
| `%H` | Hour (24-hour clock) as a zero-padded decimal number. | 00, 01, ..., 23 |
| `%-H` | Hour (24-hour clock) without zero-padding. | 0, 1, ..., 23 |
| `%I` | Hour (12-hour clock) as a zero-padded decimal number. | 01, 02, ..., 12 |
| `%-I` | Hour (12-hour clock) without zero-padding. | 1, 2, ..., 12 |
| `%p` | Locale's AM or PM. | AM, PM |
| `%M` | Minute as a zero-padded decimal number. | 00, 01, ..., 59 |
| `%-M` | Minute without zero-padding. | 0, 1, ..., 59 |
| `%S` | Second as a zero-padded decimal number. | 00, 01, ..., 59 |
| `%-S` | Second without zero-padding. | 0, 1, ..., 59 |
| `%f` | Microsecond as a decimal number, zero-padded on the left. | 000000 - 999999 |
| `%z` | UTC offset in the form `+HHMM` or `-HHMM`. | +0545 |
| `%Z` | Time zone name. | UTC |
| `%j` | Day of the year as a zero-padded decimal number. | 001, 002, ..., 366 |
| `%-j` | Day of the year as a decimal number without zero-padding. | 1, 2, ..., 366 |
| `%U` | Week number of the year, with Sunday as the first day of the week. Days before the first Sunday are in week 0. | 00, 01, ..., 53 |
| `%W` | Week number of the year, with Monday as the first day of the week. Days before the first Monday are in week 0. | 00, 01, ..., 53 |
| `%c` | Locale's appropriate date and time representation. | Mon Sep 30 07:06:05 2013 |
| `%x` | Locale's appropriate date representation. | 09/30/13 |
| `%X` | Locale's appropriate time representation. | 07:06:05 |
| `%%` | A literal `%` character. | % |
