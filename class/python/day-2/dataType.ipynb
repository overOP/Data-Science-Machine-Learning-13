1. Arithmetic Operators
Used to perform standard mathematical calculations.
+ : Addition (5 + 2 = 7)- : Subtraction (5 - 2 = 3)
* : Multiplication (5 * 2 = 10)
/ : True Division (always returns a float: 5 / 2 = 2.5)
// : Floor Division (rounds down to the nearest whole integer: 5 // 2 = 2)
% : Modulus (returns the division remainder: 5 % 2 = 1)
** : Exponentiation (power: 5 ** 2 = 25)

```python
a = 4
b = 3.4

c = a + b
print(c)
print(type(c))
```
