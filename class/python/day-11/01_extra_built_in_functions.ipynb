### map()

The `map()` function in Python applies a given function to each item of an iterable (e.g., a list or a tuple) and returns a new iterable with the transformed items.

The general syntax for `map()` is:

```python
map(function, iterable)

````python
def get_cube(num):
    return num ** 3

ls = [1,4,0,2,6,9]

map_obj = map(get_cube, ls)

map_obj
````

````python
list(map_obj)
````

In Python, map() is a built-in function that applies a function to every item in an iterable (such as a list, tuple, or set) and returns a map object (an iterator).

map(function, iterable)
map(function, iterable1, iterable2, ...)

````python
numbers = [1,2,3,4,5]

result = map(lambda x:x ** 2, numbers)

print(list(result))
````

````python
nums = ["1", "2", "3", "4"]

result = map(int, nums)

print(list(result))
````

````python
def greet(name):
    return f"Hello, {name}"

names = ["Pradeep", "Alice", "Bob", "Charlie"]

result = map(greet, names)

print(list(result))
````

````python
a = [1, 2, 3]
b = [4, 5, 6]

result = map(lambda x, y: x + y, a, b)

print(list(result))
````

````python
numbers = list(map(int, input().split()))

print(numbers)
````

filter()
The filter() function in Python is used to filter an iterable (e.g., a list or a tuple) based on a given condition. It takes two arguments: a function that specifies the condition, and an iterable to be filtered.

The function returns an iterator that contains the elements from the iterable that satisfy the condition specified by the function. The general syntax for filter() is:

filter(function, iterable)

````python
ls = [1,4,0,2,6,9]

filter_obj = filter(lambda num : num % 2 != 0, ls)

filter_obj
````

````python
list(filter_obj)
````

filter() in Python is used to select elements from an iterable based on a condition.

filter(function, iterable)

The function should return True or False.

````python
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
````

### Here:

- `x % 2 == 0` checks whether the number is even.
- If `True` → Keep the number.
- If `False` → Remove the number.

````python
def is_positive(num):
    return num > 0

numbers = [-2, -1, 0, 1, 2]

result = filter(is_positive, numbers)

print(list(result))
````

## reduce()

The `reduce()` function in Python is used to apply a given function to the items of an iterable (e.g., a list or a tuple) in a cumulative way.

It applies the function to the first two elements, then applies the function to the result and the next element, and so on, until all the items have been processed.

The general syntax for `reduce()` is:

```python
reduce(function, iterable[, initializer])

````python
from functools import reduce

def add(x, y):
    return x + y

myList = [[1, 3, 4], [7, 9, 11], [13]]

res = reduce(add, myList)

res
````

`reduce(function, iterable)`

````python
numbers = [1, 2, 3, 4, 5]

result = reduce(lambda x, y: x + y, numbers)

print(result)
````

````python
numbers = [1, 2, 3, 4]

result = reduce(lambda x, y: x * y, numbers)

print(result)
````

| Function | Purpose | Example |
|----------|---------|---------|
| `map()` | Transform every item | `[1, 2, 3] → [2, 4, 6]` |
| `filter()` | Select items | `[1, 2, 3, 4] → [2, 4]` |
| `reduce()` | Combine into one value | `[1, 2, 3, 4] → 10` |
