### Object Oriented Programming

Object-oriented programming is a programming paradigm that is centered around objects, which are instances of classes that encapsulate data and

behavior. In Python, everything is an object, including integers, strings, lists, and functions. Python's support for OOP is a key feature of

the language that makes it powerful and flexible.

# Class and Objects

A class is a blueprint or a template for creating objects, which are instances of the class.

Classes encapsulate data and the behavior that operates on that data.

A class defines a set of attributes and methods that can be used to create objects.

In Python, you can define a class using the `class` keyword.

### Key Components of a Python Class

- **Attributes** → The characteristics or properties of an object.
- **Methods** → The functions that are associated with the object.

### Syntax

```python
class ClassName:
    def method(self):
        pass

````python
class Calculator:
    def __init__(self, num1, num2):
        print("Making Objects..")
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def multiply(self, num3: int) -> int | float:
        """
        Docuentaion
        """
        return self.num1 * self.num2 * num3

calc = Calculator(5,9)
res = calc.add()
res
````
